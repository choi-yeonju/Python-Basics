## 문제 1

blood_pressure 데이터셋은 부인 15명을 대상으로 평상시 혈압을 측정한 뒤, 이들에게 이 피임약을 일정 기간 복용하게 한 후 이들의 혈압을 다시 측정한 결과를 기록한 데이터이다. 이 데이터를 통해 피임약 복용이 혈압에 영향을 주는지 분석하라.

```python
df1 = load_data('blood_pressure')
df1.head()
```

```python
my_stats.test_paired(df1, '복용전', '복용후')
```

### 💡인사이트

차이값이 정규성을 충족하여 **대응표본 t-검정**을 수행하였다. 복용 전후 혈압의 차이는 통계적으로 유의하며(t = -3.105, 양측 p = 0.008 < 0.05), 단측 검정에서도 `복용후 < 복용전`이 유의하다(p = 0.004).

즉, 피임약 복용 후 혈압이 유의하게 **감소**하였으며, 피임약 복용은 혈압에 영향(혈압을 낮추는 방향)을 준다고 할 수 있다.

## 문제 2

runner_diet 데이터셋은 고등학교 육상선수 10명에게 감량훈련(2~4kg 정도)을 시킨 후 100m 달리기를 실시한 결과를 조사한 데이터이다. 이를 통해 고등학교 육상선수에게 체중감량을 시키면 달리기에 어떤 변화가 일어나는지 조사하고자 한다.  감량 전과 후에 달리기 속도에 차이가 있는지 분석하라.

```python
df2 = load_data('runner_diet')
df2
```

```python
my_stats.test_paired(df2, '감량전', '감량후')
```

### 💡인사이트

차이값이 정규성을 충족하여 **대응표본 t-검정**을 수행하였다. 감량 전후 100m 기록의 차이는 통계적으로 유의하며(t = -2.375, 양측 p = 0.042 < 0.05), 단측 검정에서 `감량후 < 감량전`이 유의하다(p = 0.021).

즉, 체중감량 후 100m 기록이 유의하게 **단축**되었으며, 감량은 육상선수의 달리기 기록을 향상(속도 증가)시킨다고 할 수 있다.

## 문제 3

wrestler_diet 데이터셋은 고등학교 레슬링 선수에게 체중감량을 시키면 체력에 어떤 변화가 일어나는지 조사하기 위하여, 어느 고등학교의 레슬링 선수 12명을 표본추출하여 감량시킨 후 얻은 데이터이다.
감량 전과 감량 후 간에 차이가 있는가의 대응비교를 악력, 윗몸일으키기, 턱걸이에 대하여 실시하고 분석하라.

```python
df3 = load_data('wrestler_diet')
df3
```

```python
field_group = [['감량전 악력', '감량후 악력'], ['감량전 윗몸일으키기', '감량후 윗몸일으키기'], ['감량전 턱걸이', '감량후 턱걸이']]

for field in field_group:
    result = my_stats.test_paired(df3, field[0], field[1], title=field[0] + ' vs ' + field[1])
    display(result)
```

### 💡인사이트

악력·윗몸일으키기·턱걸이 세 항목 모두 감량 전후 차이가 통계적으로 유의하였다.

- **악력**: 정규성 미충족으로 Wilcoxon 검정 수행. 양측 p = 0.035 < 0.05로 유의하며 `감량후 < 감량전`(단측 p = 0.018) → 유의하게 **감소**.
- **윗몸일으키기**: 대응표본 t-검정. 양측 p < 0.001로 유의하며 `감량후 > 감량전`(단측 p < 0.001) → 유의하게 **증가**.
- **턱걸이**: 대응표본 t-검정. 양측 p < 0.001로 유의하며 `감량후 > 감량전`(단측 p < 0.001) → 유의하게 **증가**.

즉, 체중감량 후 악력(근력)은 감소한 반면, 윗몸일으키기·턱걸이(근지구력)는 향상되었다.

## 문제 4

서울시에서 정의하는 상권의 종류는 발달상권, 골목상권, 관광특구, 전통시장이 있다.

`commercial` 데이터는 서울시의 2023년 상권에 따른 분기별 매출액과 매출건수에 대한 자료이다.

주어진 데이터를 활용하여 서울시 골목상권의 2023년 상반기(1,2분기)와 하반기(3,4분기) 매출액이 어떤 차이를 보이는지 분석하라.

주어진 데이터는 서울시에서 제공하는 집계가 완료된 데이터이므로 모든 결측치는 없는 상태이며 이상치도 실제 매출액 데이터이므로 별도의 정제 과정 없이 진행하는 것으로 한다.

```python
df4 = load_data('commercial')
df4.head()
```

```python
# 결측치 확인
df4.isna().sum()
```

```python
# 상권유형 확인
df4['type'].unique()
```

```python
# 상・하반기 구분 파생변수 추가
df4['구분'] = df4['quarter'].map({1: '상반기', 2: '상반기', 3:'하반기', 4:'하반기'})
df4.head()
```

```python
# 골목 상권 데이터만 추출
df5 = df4[df4['type'] == '골목상권']
df5.head()
```

```python
# 상・하반기 매출액에 대한 Wide 형식의 피벗 테이블 생성
df6 = pivot_table(df5, index='code', columns='구분', values='sales_amount')
df6
```

```python
my_stats.test_paired(df6, '상반기', '하반기', title='골목상권 상반기 vs 하반기 매출액')
```

### 💡인사이트

상·하반기 매출액의 차이값이 정규성을 충족하지 않아 **Wilcoxon 부호순위 검정**을 수행하였다. 골목상권의 상·하반기 매출액 차이는 통계적으로 매우 유의하며(양측 p ≈ 1.6e-94 < 0.05), 단측 검정에서 `하반기 > 상반기`가 유의하다(p < 0.001).

즉, 서울시 골목상권의 2023년 하반기 매출액은 상반기보다 유의하게 **높다**.
