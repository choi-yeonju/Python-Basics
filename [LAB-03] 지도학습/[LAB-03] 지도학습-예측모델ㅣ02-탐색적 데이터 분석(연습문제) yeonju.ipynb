```python
target = "price" # 종속변수 이름    
target_is_continuous = True # 종속변수 연속형 여부 

# 분석 대상에서 제외할 변수
# -> 식별자(ID), 다른 변수와 중복되는 항목, 일반 변수로 다룰 수 없는 값 등을 나열한다
# -> 이번 데이터셋에서는 위경도가 여기에 해당한다
# 거래날짜, 우편번호, 위경도, 리모델링 연도 제외 
exclude_cols = ["date", "zipcode", "lat", "long", "yr_renovated"]

nominal_cols = my_qtcheck.get_categorical_column_names(df) # 범주형 변수
continuous_cols = my_qtcheck.get_number_column_names(df) # 연속형 변수

if target_is_continuous:           # 종속변수가 연속형 변수라면 ?
    continuous_cols.remove(target) # 연속형 변수 이름에서 종속변수 항목 제거 
else:                              # 종속변수가 범주형 변수라면?
    nominal_cols.remove(target)    # 범주형 변수 이름에서 종속변수 항목 제거 

# 제외 대상으로 지정한 변수를 각 목록에서 제거한다 
# -> 범주형에도 적용하므로 제외 대상이 범주형인 경우에도 그대로 동작한다. 
continuous_cols = [c for c in continuous_cols if c not in exclude_cols] 
nominal_cols = [c for c in nominal_cols if c not in exclude_cols]   

print("종속변수: ", target)
print("종속변수 유형: " + ("연속형" if target_is_continuous else "범주형"))
print("범주형: ", nominal_cols)
print("연속형: ", continuous_cols)
print("제외: ", exclude_cols)
```

```python
# #### 3. 캘리포니아 EDA 범위표에서 대상이 없어 '미적용'으로 비어 있던 두 집단 비교를, 이번에는 워터프론트 여부를 두고 수행하게 됩니다. 정규성 점검 결과에 따라 채택된 검정의 이름은 무엇인가요? (검정 이름 1개)

# - **정답**: `Mann-Whitney U test`
# - **복습 개념**: 두 독립집단 비교에서의 검정 자동 선택입니다. 두 집단이 모두 정규성을 충족하면 등분산 여부에 따라 Student 또는 Welch t검정을, 하나라도 충족하지 못하면 비모수 검정을 씁니다.
# - **풀이 접근**: 명목형 변수 중 고유값이 2개인 것만 골라내고, 집단별 가격을 열로 나눈 형태로 바꾼 뒤 두 집단 비교를 수행합니다. 결과표의 검정 이름 열을 읽으면 어떤 절차가 선택되었는지 그대로 나옵니다.
# - **근거(계산 결과)**: 결과표에 **Mann-Whitney U test**로 찍힙니다. 가격의 왜도가 4.024여서 두 집단 모두 정규성을 충족하지 못했기 때문입니다. 양측검정 p는 0.000으로 유의하고, 워터프론트인 주택 163채의 평균은 1,661,876달러로 나머지 21,450채의 531,564달러보다 **3배 이상** 높습니다.
# # - **실무 포인트**: "두 집단 비교 = t검정"으로 외워 두면 여기서 어긋납니다. 검정 이름은 데이터의 가정 점검 결과가 정하는 것이지 집단 수가 정하는 것이 아닙니다. 상관분석에서 Pearson 대신 Spearman이 선택된 것과 같은 원리입니다. 다만 이 결과를 그대로 믿기 전에 표본을 보세요. 워터프론트 주택은 **163채(0.8%)** 뿐입니다. 캘리포니아의 ISLAND 5건만큼 극단적이지는 않지만, 더미 인코딩 후 계수 해석에는 주의가 필요합니다.
```

```python
for n in nominal_cols:
    # 3개 이상의 고유값이 존재하는 명목형 독립변수에 대해서만 검정을 수행한다 
    if len(df[n].unique()) >= 3:
        print() # 빈 줄 출력용 

        # 분석 단위 제목 출력 
        display(Markdown(f"#### -> {n}({column_means[n]}) -> {target}"))    

        # origin이 아닌 df를 사용해야 서열척도로 지정한 범주 순서가 반영된다 
        display(my_stats.anova_oneway(df, y=target, between=n)) 

        # 사후검정 
        display(my_stats.posthoc_oneway(df, y=target, between=n, plot=False))   
```

```python
# #### 4. 3집단 이상인 명목형 변수 두 종에 대해 집단 간 평균 비교를 수행했더니 둘 다 유의확률이 0.001보다 작게 나왔습니다. 그런데 그중 하나는 효과크기가 '무시할 수준'으로 판정되었습니다. 그 변수는 무엇인가요? (변수명 1개)

# - **정답**: `condition`
# - **복습 개념**: 유의성과 효과크기의 구분입니다. 유의확률은 "차이가 있느냐"에, 효과크기(np2)는 "그 차이가 얼마나 크냐"에 답합니다. 표본이 크면 아주 작은 차이도 유의하게 나옵니다.
# - **풀이 접근**: 고유값이 3개 이상인 명목형 변수만 골라 집단 간 평균 비교를 수행하고, 결과표의 효과크기 열을 서로 비교합니다. 이어서 사후검정으로 어떤 쌍이 실제로 갈리는지 확인합니다.
# - **근거(계산 결과)**: **condition**은 F=52.777로 p가 0.000이지만 np2=**0.007**이라 판정이 `Negligible`입니다. 반면 view는 np2=0.168로 `Large`입니다. condition의 집단 평균을 보면 3등급 542,013달러, 4등급 521,200달러로 **등급이 높은 쪽이 오히려 싸고**, 사후검정에서도 3 대 4 쌍의 효과크기가 Negligible입니다. 21,613건이라는 표본 크기가 2만 달러 차이를 유의하게 만든 것뿐입니다.
# # - **자주 하는 실수**: p가 0.000이라는 이유만으로 채택 목록에 올리기 쉽습니다. 캘리포니아의 해안권 3종이 "유의하지만 효과크기 Negligible"이라 병합 대상이 되었던 것과 같은 상황입니다. **표본이 크면 p는 거의 항상 작아지므로, 채택 여부는 효과크기로 판단하세요.** 반대로 view는 조망 0점 496,564달러 대 4점 1,463,711달러로 100만 달러 가까이 벌어지는 진짜 설명변수입니다.
```
