### 💻 코드 작성

#### 데이터 가져오기 (품질 점검 결과 반영)

```python
origin = load_data("kc_house")

# 지난 단원의 품질 점검대로 중복 행만 정리하고, 값은 원본 그대로 사용한다.
df0 = my_qtcheck.check_duplicates(origin)

print(f"데이터 크기: {df0.shape}")
df0.head()
```

#### 타입 변환 (범주형 확정)

```python
# 지난 단원에서 확정한 범주형 2종(워터프론트 여부·우편번호)에
# 이번 단원에서 집단으로 볼 점수 2종(조망 점수·주택 상태 점수)을 더한다.
df = my_qtcheck.set_type(df0, as_category=["waterfront", "zipcode", "view", "condition"])
```

#### 변수 유형 분류

```python
target = "price"              # 종속변수 이름
target_is_continuous = True   # 종속변수 연속형 여부

# 분석 대상에서 제외할 변수
# -> 거래 시점·좌표, 지역이 70개라 집단 비교가 어려운 우편번호,
#    그리고 "연도가 아니라 여부"로 바꿔 써야 하는 리모델링 연도
exclude_cols = ["date", "zipcode", "lat", "long", "yr_renovated"]

nominal_cols = my_qtcheck.get_categorical_column_names(df)   # 범주형 변수
continuous_cols = my_qtcheck.get_number_column_names(df)     # 연속형 변수

if target_is_continuous:               # 종속변수가 연속형 변수라면?
    continuous_cols.remove(target)     # 연속형 변수 이름에서 종속변수 항목 제거
else:                                  # 종속변수가 범주형 변수라면?
    nominal_cols.remove(target)        # 범주형 변수 이름에서 종속변수 항목 제거

# 제외 대상으로 지정한 변수를 각 목록에서 제거한다.
continuous_cols = [c for c in continuous_cols if c not in exclude_cols]
nominal_cols = [c for c in nominal_cols if c not in exclude_cols]

print("종속변수 :", target)
print("범주형   :", nominal_cols)
print("연속형   :", continuous_cols)
print("제외     :", exclude_cols)
```

#### 컬럼 의미를 정리한 딕셔너리

```python
column_means = {
    "price":         "주택 가격 (USD)",
    "bedrooms":      "침실 수",
    "bathrooms":     "욕실 수",
    "sqft_living":   "거주 공간 면적",
    "sqft_lot":      "대지 면적",
    "floors":        "층 수",
    "waterfront":    "워터프론트 여부",
    "view":          "조망 점수",
    "condition":     "주택 상태 점수",
    "grade":         "건축 등급 점수",
    "sqft_above":    "지상 면적",
    "sqft_basement": "지하 면적",
    "yr_built":      "건축 연도",
    "sqft_living15": "인근 15채의 평균 거주 면적",
    "sqft_lot15":    "인근 15채의 평균 대지 면적"
}
```

#### 단변량 분석 - 연속형 (종속변수 + 독립변수)

```python
num_desc = my_qtcheck.numerical_summary(df[[target] + continuous_cols])
num_desc[['mean', '50%', 'skew', 'kurt', 'outliers_ratio', 'log_need']]
```

#### 단변량 분석 - 명목형

```python
cat_desc = my_qtcheck.categorical_summary(df[nominal_cols])

for n in nominal_cols:
    display(Markdown(f"#### ▶︎ {n}({column_means[n]}) 분포"))

    freq = df[n].value_counts().rename("빈도").to_frame()
    freq["비율"] = df[n].value_counts(normalize=True)
    display(freq)
```

#### 이변량 분석 - 상관분석 (연속형 → 연속형 종속변수)

```python
corr_list = []

for f in continuous_cols:
    corr_list.append(my_stats.correlation(df, x=f, y=target, plot=False))

# 11개 결과를 한 표로 합쳐 상관계수 내림차순 정렬
corr_result = concat(corr_list).sort_values("coef", ascending=False)
display(corr_result)

print("[ 상관 강도 판정별 변수 개수 ]")
print(corr_result['strength'].value_counts())
```

#### 이변량 분석 - T검정 (2집단 명목형 → 종속변수)

```python
for n in nominal_cols:
    # 2개의 고유값만 존재하는 범주형 독립변수에 대해서만 검정을 수행한다.
    if len(df[n].unique()) == 2:
        display(Markdown(f"#### ▶︎ {n}({column_means[n]}) → {target}"))

        # 수집된 데이터가 long 타입이므로 검정 대상만 wide 타입으로 변환해야 한다.
        wide_df = my_prep.long2wide(df, hue=n, values=target)
        display(my_stats.test_independent(wide_df,
                    group1=wide_df.columns[0], group2=wide_df.columns[1], plot=False))

        display(df.groupby(n, observed=True)[target].agg(['count', 'mean']))
```

#### 이변량 분석 - ANOVA (3집단 이상 명목형 → 종속변수)

```python
for n in nominal_cols:
    # 3개 이상의 고유값이 존재하는 명목형 독립변수에 대해서만 검정을 수행한다.
    if len(df[n].unique()) >= 3:
        display(Markdown(f"#### ▶︎ {n}({column_means[n]}) → {target}"))

        display(my_stats.anova_oneway(df, y=target, between=n))

        # 사후검정
        display(my_stats.posthoc_oneway(df, y=target, between=n, plot=False))
```

#### 다변량 분석 - 독립변수간 상관 분석

```python
corr = my_stats.multi_correlation(df, columns=continuous_cols, plot=False)
my_stats.correlation_summary(corr)
```

#### 2. 캘리포니아에서는 연속형 독립변수 6종 가운데 상관 강도가 '보통' 이상으로 판정된 것이 중위 소득 하나뿐이었습니다. 이 자료의 연속형 독립변수 11종 중 '보통'으로 판정된 변수는 몇 개인가요? (정수, 단위: 개)

- **정답**: `7`
- **복습 개념**: 상관분석 결과표의 강도 판정입니다. 상관계수의 절댓값이 0.7을 넘으면 강함, 0.3 초과 0.7 이하면 보통, 0 초과 0.3 이하면 약함으로 라벨이 붙습니다.
- **풀이 접근**: 연속형 독립변수 이름을 하나씩 돌면서 종속변수와의 상관분석을 수행하고, 결과를 한 표로 합칩니다. 그 표의 강도 열만 뽑아 값별로 개수를 셉니다. 참고로 11종이라는 숫자는 지난 단원에서 값의 범위를 점검했던 15개에서 종속변수(price)와 이번에 범주형으로 돌린 두 종(view·condition), 그리고 제외한 리모델링 연도를 뺀 결과입니다.
- **근거(계산 결과)**: 판정 결과는 `Moderate` **7개**(grade 0.658 / sqft_living 0.644 / sqft_living15 0.572 / sqft_above 0.542 / bathrooms 0.497 / bedrooms 0.345 / floors 0.322), `Weak` 4개(sqft_basement 0.252 / yr_built 0.102 / sqft_lot 0.075 / sqft_lot15 0.063)입니다. `Strong`은 0개이며, 11종 모두 p가 0.000이라 유의성 판정은 전부 True입니다.
- **함께 생각해 볼 점**: 캘리포니아는 구역 단위로 집계된 자료라 집 한 채의 특성이 평균에 묻혀 상관이 잘 잡히지 않았습니다. 이 자료는 주택 한 채가 한 행이라 면적·등급·욕실 수가 가격과 곧바로 맞물립니다. **설명변수가 많아진 것이 아니라 집계 단위가 달라진 것입니다.** 다만 쓸 만한 변수가 7개라는 뜻은 아닙니다. 이 7종이 서로 겹치는지는 다변량 단계에서 따로 확인해야 합니다.

#### 3. 캘리포니아 EDA 범위표에서 대상이 없어 '미적용'으로 비어 있던 두 집단 비교를, 이번에는 워터프론트 여부를 두고 수행하게 됩니다. 정규성 점검 결과에 따라 채택된 검정의 이름은 무엇인가요? (검정 이름 1개)

- **정답**: `Mann-Whitney U test`
- **복습 개념**: 두 독립집단 비교에서의 검정 자동 선택입니다. 두 집단이 모두 정규성을 충족하면 등분산 여부에 따라 Student 또는 Welch t검정을, 하나라도 충족하지 못하면 비모수 검정을 씁니다.
- **풀이 접근**: 명목형 변수 중 고유값이 2개인 것만 골라내고, 집단별 가격을 열로 나눈 형태로 바꾼 뒤 두 집단 비교를 수행합니다. 결과표의 검정 이름 열을 읽으면 어떤 절차가 선택되었는지 그대로 나옵니다.
- **근거(계산 결과)**: 결과표에 **Mann-Whitney U test**로 찍힙니다. 가격의 왜도가 4.024여서 두 집단 모두 정규성을 충족하지 못했기 때문입니다. 양측검정 p는 0.000으로 유의하고, 워터프론트인 주택 163채의 평균은 1,661,876달러로 나머지 21,450채의 531,564달러보다 **3배 이상** 높습니다.
- **실무 포인트**: "두 집단 비교 = t검정"으로 외워 두면 여기서 어긋납니다. 검정 이름은 데이터의 가정 점검 결과가 정하는 것이지 집단 수가 정하는 것이 아닙니다. 상관분석에서 Pearson 대신 Spearman이 선택된 것과 같은 원리입니다. 다만 이 결과를 그대로 믿기 전에 표본을 보세요. 워터프론트 주택은 **163채(0.8%)** 뿐입니다. 캘리포니아의 ISLAND 5건만큼 극단적이지는 않지만, 더미 인코딩 후 계수 해석에는 주의가 필요합니다.

#### 4. 3집단 이상인 명목형 변수 두 종에 대해 집단 간 평균 비교를 수행했더니 둘 다 유의확률이 0.001보다 작게 나왔습니다. 그런데 그중 하나는 효과크기가 '무시할 수준'으로 판정되었습니다. 그 변수는 무엇인가요? (변수명 1개)

- **정답**: `condition`
- **복습 개념**: 유의성과 효과크기의 구분입니다. 유의확률은 "차이가 있느냐"에, 효과크기(np2)는 "그 차이가 얼마나 크냐"에 답합니다. 표본이 크면 아주 작은 차이도 유의하게 나옵니다.
- **풀이 접근**: 고유값이 3개 이상인 명목형 변수만 골라 집단 간 평균 비교를 수행하고, 결과표의 효과크기 열을 서로 비교합니다. 이어서 사후검정으로 어떤 쌍이 실제로 갈리는지 확인합니다.
- **근거(계산 결과)**: **condition**은 F=52.777로 p가 0.000이지만 np2=**0.007**이라 판정이 `Negligible`입니다. 반면 view는 np2=0.168로 `Large`입니다. condition의 집단 평균을 보면 3등급 542,013달러, 4등급 521,200달러로 **등급이 높은 쪽이 오히려 싸고**, 사후검정에서도 3 대 4 쌍의 효과크기가 Negligible입니다. 21,613건이라는 표본 크기가 2만 달러 차이를 유의하게 만든 것뿐입니다.

#### 5. 이제 팀장에게 1순위 변수를 보고해야 합니다. 종속변수와의 상관이 가장 강했고, 독립변수끼리 겹치는 덩어리 안에서도 대표로 남길 만한 변수는 무엇인가요? (변수명 1개)

- **정답**: `grade`
- **복습 개념**: 다변량 단계의 중복 점검과 최종 변수 선택입니다. 이변량에서 살아남은 변수라도 서로 강하게 겹치면 하나만 남기고, 그때 기준은 종속변수와의 설명력입니다.
- **풀이 접근**: 연속형 독립변수 전체의 상관행렬을 만들고, 강한 상관으로 묶인 변수 덩어리를 요약합니다. 그 덩어리 안에서 이변량 결과표의 상관계수를 비교해 대표를 고릅니다.
- **근거(계산 결과)**: **grade**의 상관계수가 0.658로 11종 중 1위입니다. 중복 점검에서 grade는 sqft_living(0.716)·sqft_above와 한 덩어리로 묶이는데, 그 덩어리에서 상관계수가 가장 높은 쪽이 grade(0.658 > sqft_living 0.644 > sqft_above 0.542)입니다. 규모 축의 허브인 sqft_living은 무려 4개 변수(sqft_above 0.844 / sqft_living15 0.747 / bathrooms 0.746 / grade 0.716)와 강하게 겹칩니다.
- **결론**: 팀장에게 드릴 답은 이렇게 정리됩니다. **1순위는 `grade`, 2순위는 명목형에서 효과크기가 가장 큰 `view`입니다.** 규모 축(sqft_living·sqft_above·sqft_living15·bathrooms)은 서로 0.7 이상으로 겹치므로 grade 하나로 대표하고, sqft_lot과 sqft_lot15는 서로 0.922로 겹치면서 가격과의 상관은 0.07 수준이라 함께 제외합니다. condition은 유의하지만 효과크기가 없어 보류입니다. 캘리포니아에서는 위치(ocean_proximity)와 소득이 답이었는데, 킹 카운티에서는 **건축 등급과 조망**이 답입니다. 절차는 똑같았지만 절차가 가리킨 변수는 데이터마다 달랐습니다.
