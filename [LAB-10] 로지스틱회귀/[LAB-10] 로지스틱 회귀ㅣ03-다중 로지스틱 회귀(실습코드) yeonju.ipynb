## #01. 준비작업 

```python
# 순서 정리 (LAB-10 다중 로지스틱 회귀 - pima_indians_diabetes)

# 1. 준비작업
# 라이브러리 참조 (jussam.load_data, helpers)
# 데이터 불러오기 (load_data('pima_indians_diabetes'))

# 2. 데이터 품질검사
# 자료형 점검 (my_qtcheck.set_type) — Outcome을 category로 지정
# 중복검사 (check_duplicates)
# 결측치 확인 (check_missing_values)
# 기술통계량 확인 (numerical_summary)

# 3. "0은 정말 0인가" 판단
# Pregnancies, Age 제외한 연속형 컬럼 → 0을 NaN으로 변환 (np.where)
# → 생물학적으로 0이 불가능한 값이 결측으로 잘못 입력된 걸로 봐야 함

# 4. 결측치 처리 전략 분기
# 결측 비중 과도(SkinThickness, Insulin) → 중앙값 대체시 분산 왜곡 → 컬럼 자체 제거
# 나머지 결측 비중 낮은 변수 → 행 단위 제거 (dropna)
# → 컬럼 제거를 먼저 해야 불필요한 행 손실을 막을 수 있음 (순서 중요)

# 5. 기술통계량 재확인
# 값이 바뀌었으니 numerical_summary 재실행
# → BloodPressure 최솟값 24는 이례적이지만, "0만 결측으로 본다"는 원칙 유지하고 그대로 진행

# 6. 다중공선성 점검
# my_prep.reduce_vif(df4) → 제거된 변수 없음 (공선성 위험 없는 데이터로 판단)

# 7. 후진소거법 적용 다중 로지스틱 회귀
# 기준선: 전체 6개 변수 투입 → my_logit.fit_model, report_variables
# → 최대p값 0.301(BloodPressure) 제거
# 1단계: 5개 변수 → 최대p값 0.128(Age) 제거
#   ※ BloodPressure 제거 후 Age의 p값이 0.085→0.128로 상승
#     → 변수 하나씩만 제거해야 하는 이유가 여기서 확인됨
# 2단계: 4개 변수 → 전부 p<0.05 → 소거 종료

# 8. 결과 보고
# 모형 적합도 먼저 확인 (report_fitness) — Pseudo R²는 %로 해석 금지, 상대비교 용도만
# 개별 변수(오즈비 등) 보고 (report_variables, report_variables_text)
# → 오즈비 크기로 변수 간 영향력 직접 비교는 불가

# 9. 모듈화
# 위 전체 과정을 auto_logit(df5, y='Outcome', backward=True) 한 줄로 재현
```

### 1. 패키지 참조 

```python
from jussam import load_data
from helpers import *

import numpy as np
from IPython.display import display, Markdown
```

### 2. 데이터 가져오기 

```python
origin = load_data('pima_indians_diabetes')
origin.head()
```

## #02. 데이터 품질 검사 

### 1. 자료형 점검 

```python
df1 = my_qtcheck.set_type(origin, as_category=['Outcome'])
```

### 2. 중복 검사 및 중복 데이터 삭제 

```python
df2 = my_qtcheck.check_duplicates(df1)
```

### 3. 결측치 확인 

```python
my_qtcheck.check_missing_values(df2)
```

### 4. 연속형 변수의 기술통계량 확인 

```python
desc = my_qtcheck.numerical_summary(df2)
desc
```

### 데이터 품질 검사 : 0은 정말 0인가 

### 5. 0값을 결측치로 변경후 결측치 재확인 

```python
df3 = df2.copy()

# 연속형 컬럼의 이름을 가져와서 반복 순환 
for col in df3.select_dtypes(include='number').columns:
    # 컬럼 이름 0값을 포함하는 컬럼인 경우만 처리 
    if col not in ['Pregnancies', 'Age']:
        df3[col] = np.where(df3[col]>0, df3[col], np.nan)

my_qtcheck.check_missing_values(df3)
```

```python
# 결측치 결과 
    # 결측치가 5% 이내로 존재( 행단위 삭제 가능 )

    # 결측치가 과도하게 많음(변수 자체 제거 고려)
        # 절반 가까이를 중앙값 하나로 대체하면 분산이 인위적으로 줄어 변수가 가진 정보가 사라진다.
        # SkinThickness 
        # Insulin 
```

```python
# 결측치 비중이 과도하게 높은 컬럼 제거 
df4 = df3.drop(columns=['SkinThickness', 'Insulin'])

# 컬럼이 제거된 데이터프레임에서 결측치 비중이 낮은 변수에 대해 행 단위 제거 
df4 = df4.dropna().reset_index(drop=True)

# 결측치 제거 결과 확인 
my_qtcheck.check_missing_values(df4)
```

### 기술 통계량 재확인
값이 변경되었으므로 기술 통계량을 다시 산정해야 한다 

```python
desc = my_qtcheck.numerical_summary(df4)
desc
```

## # 후진 소거법 적용 다중 로지스틱 회귀 
### 모든 변수로 시작해서 유의하지 않은 변수를 단계별로 제거 
모든 변수로 시작 -> p값이 가장 큰 변수 하나 제거 -> 재적합 -> 반복

### 이 예제에서 제거 과정 
단계 변수수 모델투입변수 최대p값 조치\
기준선 6                0.301 BloodPressure제거\
1단계  5                0.128 Age제거\
2단계 4                 <0.05 소거종료 

## #04. 다중 로지스틱 회귀 

### 1. 기준선 데이터를 사용하여 모델 적합 
먼저 모든 변수를 넣고 기준선 모형을 적합하고, 유의확률로 걸러낼 후보를 찾는다. 

```python
fit1 = my_logit.fit_model(df5, y='Outcome')
my_logit.report_variables(fit1, df5)
```

### 유의수준 0.05 초과 변수는 BooldPressure, Age
유의하지 않다고 무의미한 변수는 아님\
다른 변수를 통제한 상태에서 추가로 설명하는 부분이 없다는 뜻\
어느 것부터, 어떻게 뺄 것인가 -> 다음 단위(후진 소거법)에서 처리 

### 2. 후진 소거 1단계 - BloodPressure 제거 
유의하지 않은 변수를 한꺼번에 빼지 않고,\
유의확률이 가장 큰 BloodPressure를 먼저 제거한 뒤 모형을 재적합 한다. \
변수 하나를 빼면 남은 변수의 계수와 표준오차가 함께 달라지므로,\
반드시 하나씩 제거하고 결과를 다시 확인해야 한다. 

```python
df6 = df5.drop(columns=['BloodPressure'])

fit2 = my_logit.fit_model(df6, y='Outcome')
my_logit.report_variables(fit2, df6)
```

## #04. 다중 로지스틱 회귀 

### 2. 후진 소거 2단계 - Age 제거
1단계 소거 후, 모델에서 유의확률이 0.05를 넘는 변수는 Age(p=0.128) 하나 이므로, \
이를 제거하고 모형을 재적합 한다. 

BloodPressure를 제거하자, Age의 유의확률이 0.085 -> 0.128로 커졌다.\
변수를 한번에 하나씩만 제거애햐 하는 이유가 여기서 드러난다. 

```python
df7 = df6.drop(columns=['Age'])

fit3 = my_logit.fit_model(df7, y='Outcome')
my_logit.report_variables(fit3, df7)
```

## #05. 결과 보고

```python
Markdown(my_logit.report_fitness(fit3))
```

### Pseudo R^2 읽는법
