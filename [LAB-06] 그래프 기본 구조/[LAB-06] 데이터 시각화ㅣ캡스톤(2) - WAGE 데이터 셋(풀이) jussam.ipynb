## 문제 11. 연령대별 평균 임금이 가장 높은 구간은?

① 20대 이하  ② 30대  ③ 40대  ④ 60대 이상

> 💡 **힌트**: 산점도(lmplot) 또는 연령대 구간화 후 막대그래프
> 🎯 **ADsP 개념**: 연속형 변수의 구간화(binning) · 비선형 추세

### 정답 및 해설

**정답: ③ 40대**

연령대별 평균 임금은 ~30세 89.5, 30대 113.3, 40대 118.6, 50대 118.2, 60대 이상 112.8로 **40대가 정점**이며 이후 완만히 감소합니다.

```python
# 연령대별 평균 임금 → 나이를 구간화한 뒤 막대그래프
df = origin.copy()
df["age_group"] = pd.cut(df["age"], bins=[17, 30, 40, 50, 60, 80],
                         labels=["~30", "31~40", "41~50", "51~60", "61~"])

my_plot.barplot(data=df, x="age_group", y="wage",
                title="연령대별 평균 임금", xlabel="연령대", ylabel="평균 임금(wage)")
```
