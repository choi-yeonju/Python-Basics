## #02. Scatter Plot 시각화

### 1. 기본 사용 방법 - 팽귄의 날개 길이와 몸무게의 관계

- 데이터 프레임, 그래프의 x축에 적용할 변수명, y축에 적용할 변수명을 지정한다.

```python
# 1) 그래프 초기화
my_plot.init(title="팽귄의 날개 길이와 몸무게의 관계", xlabel="날개 길이", ylabel="몸무게")

# 2) Scatter Plot 그리기
# 패턴1 - 자동인식 패턴 => 모든 컬럼에 대해 작용하므로 권장하지 않음
#sb.scatterplot(data=origin)
# 패턴2 - 개별 데이터 직접 지정 패턴 => 사용 가능
#sb.scatterplot(x=origin['flipper_length_mm'], y=origin['body_mass_g'])
# 패턴3 - DataFrame + 컬럼명 패턴 => 가장 권장하는 패턴
sb.scatterplot(data=origin, x='flipper_length_mm', y='body_mass_g')

# 3) 그래프 출력
my_plot.show()
```

#### 💡 인사이트

- 📈 **양의 관계가 뚜렷함**: 날개 길이가 길어질수록 몸무게가 커지는 경향이 산점도에서 명확히 보임.
- 📍 **데이터 분포 구간이 넓음**:  
  - 날개 길이: 대략 170~230  
  - 몸무게: 대략 2700~6300
- 🔹 **중간 구간 밀집**: 날개 길이 약 185~200, 몸무게 약 3200~4300 구간에 점이 많이 모여 있음.
- 🔹 **상위 구간 밀집**: 날개 길이 약 210~225, 몸무게 약 4700~5700 구간에도 점이 집중되어 보임.
- 🧩 **같은 날개 길이에서도 몸무게 분산 존재**: 동일하거나 유사한 날개 길이에서 몸무게 값이 여러 수준으로 퍼져 있어, 날개 길이 외 요인도 함께 작용함을 시사함.

### 2. Scatter Plot의 옵션 - 팽귄의 날개 길이와 몸무게의 관계

```python
# 1) 그래프 초기화
my_plot.init(title="팽귄의 날개 길이와 몸무게의 관계", xlabel="날개 길이", ylabel="몸무게")

# 2) Scatter Plot 그리기
sb.scatterplot(data=origin, x='flipper_length_mm', y='body_mass_g',
               marker="^",          # 마커 모양
               color="red",         # 마커 색상
               s=150,               # 마커 크기 (기본값=100)
               edgecolor="black",   # 마커 테두리 색상
               linewidth=2,         # 마커 테두리 두께
               alpha=0.7)           # 마커 투명도 (1=불투명, 0=투명)

# 3) 그래프 출력
my_plot.show()
```

### 3. 군집별 분류 - 팽귄 종에 따른 길이와 부리 두께의 관계

- hue 파라미터에 카테고리 컬럼을 지정한다.
- hue 파라미터를 적용하는 경우에 한해서 palette 파라미터를 사용할 수 있다.

```python
# 1) 그래프 초기화
my_plot.init(title="팽귄 종에 따른 부리 길이와 부리 두께의 관계", 
             xlabel="부리 길이", ylabel="부리 두께")

# 2) Scatter Plot 그리기
sb.scatterplot(data=origin, x='bill_length_mm', y='bill_depth_mm',
               hue="species",       # 마커 색상을 종(species)별로 다르게 설정
               palette="tab10",     # 색상 팔레트 설정
               marker="o",          # 마커 모양
               s=100,               # 마커 크기 (기본값=100)
               edgecolor="#ffffff", # 마커 테두리 색상
               linewidth=1.5,       # 마커 테두리 두께
               alpha=1)             # 마커 투명도 (1=불투명, 0=투명)

# 3) 그래프 출력
my_plot.show()
```

#### 💡 인사이트

- 🟦 **종별 군집이 뚜렷하게 구분됨**: Adelie, Chinstrap, Gentoo가 부리 길이와 부리 두께 기준으로 서로 다른 영역에 모여 있음.
- 📏 **Adelie는 짧고 두꺼운 편**: 부리 길이가 비교적 짧고, 부리 두께는 17~20 근처에 많이 분포함.
- 📐 **Chinstrap은 길고 두꺼운 편**: 부리 길이가 길고, 부리 두께도 18~20대에 많이 분포함.
- 🟩 **Gentoo는 길지만 상대적으로 얇음**: 부리 길이는 중간~긴 구간에 분포하지만, 부리 두께는 13~16 근처에 집중됨.
- 🔍 **겹침 구간도 존재함**: 일부 구간에서 종 간 점들이 섞여 있어, 부리 길이와 두께만으로도 상당 부분 구분되지만 완전 분리는 아님.

### 2. 팔레트 적용

- palette를 적용할 수 있다.

```python
my_plot.scatterplot(data=origin, x='bill_length_mm', 
            y='bill_depth_mm', hue="species", palette="Greens")
```
