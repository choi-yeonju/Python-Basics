## 문제풀이

### 문제 1

권역별로 녹지율이 높아질수록 평균월세는 어떻게 되는가? (양/음)

- **정답**: 양(녹지율이 높을수록 평균월세도 높아진다 = 양의 상관관계)
- **해설**: 회귀선이 우상향(↗)한다는 것은 x축(녹지율)이 커질 때 y축(평균월세)도 함께 커진다는 뜻이므로 **양의 상관관계**이다. 같은 권역 안에서는 녹지가 많은 쾌적한 동네일수록 월세가 비싸다.

### 문제 2

도시규모에 따라 권역별로 녹지율과 평균 월세를 시각화 했을 때 어떤 규모의 평균 월세가 더 높다고 볼 수 있는가?

- **정답**: 대도시
- **해설**: 도시규모별 평균월세는 대도시 약 96.8만원, 중소도시 약 86.9만원으로 대도시가 더 높다. 그래서 모든 권역에서 대도시 패싯의 회귀선이 중소도시보다 위쪽에 그려진다.
