## #03. 문제 풀이

### 문제 1

A학과와 C학과 중 평균 점수가 더 높은 학과는?

- **정답**: C학과
- **해설**: 그래프에 표시된 Mean 값을 비교하면 A학과는 43.75점, C학과는 56.17점으로 C학과의 평균이 약 12점 더 높다.

### 문제 2

C학과의 점수 분포에서 관찰되는 봉우리(peak)의 개수는?

- **정답**: 2개 (이봉형 분포)
- **해설**: C학과의 KDE는 50점 부근의 큰 봉우리와 85점 부근의 작은 봉우리, 두 개의 정점이 나타난다. 이는 학과 내에 서로 다른 점수대를 가진 두 집단이 존재할 가능성을 시사한다.

### 문제 3

A학과와 C학과 중 점수가 더 넓게 퍼져 있는(산포가 큰) 학과는?

- **정답**: A학과
- **해설**: A학과의 분포는 약 0점부터 100점까지 넓게 퍼진 완만한 모양인 반면, C학과의 분포는 40~60점 구간에 좁고 뾰족하게 집중되어 있다. 즉, A학과의 학생 간 점수 편차가 더 크다.

### 문제 4

C학과의 KDE에서 평균(56.17)이 최빈값(약 50점)보다 큰 위치에 있다. 이로부터 알 수 있는 분포의 왜도(skewness)는?

- **정답**: 양의 왜도 (오른쪽 꼬리 분포)
- **해설**: 85점 부근의 두 번째 봉우리(고득점 집단)가 평균을 오른쪽으로 끌어올리기 때문에 평균 > 최빈값 관계가 성립한다. 분포가 오른쪽으로 긴 꼬리를 가지는 양의 왜도(positively skewed) 형태이다.

### 문제 5

두 학과 중 '평균값'을 성적의 대표값으로 사용하기에 더 적합한 학과는?

- **정답**: A학과
- **해설**: A학과는 단봉형의 좌우 대칭에 가까운 분포로, 평균(43.75)이 분포의 중심과 다수 학생이 위치한 점수대를 잘 대표한다. 반면 C학과는 이봉형 분포이므로 평균(56.17)이 두 봉우리 사이의 골짜기(저밀도 구간)에 놓여, 실제로는 그 점수대에 학생이 적어 어떤 집단도 대표하지 못한다.
