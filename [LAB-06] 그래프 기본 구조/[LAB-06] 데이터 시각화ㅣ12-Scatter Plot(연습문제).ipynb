이 데이터를 Scatter Plot으로 시각화하여 데이터에 숨어 있는 핵심 인사이트를 찾아내는 연습을 수행합니다.

## #02. Scatter Plot 시각화

```python
my_plot.scatterplot(
    data=origin, x='study_hours', y='exam_score', hue='group', 
    palette="Set1", marker="^", size=120, 
    edgecolor="black", linewidth=1.5, alpha=0.8, 
    title="학습 방식별 공부 시간과 시험점수의 관계", 
    xlabel="공부 시간", ylabel="시험 점수")
```

## 문제 풀이

### 문제 1

산점도에서 `study_hours`(공부 시간)가 증가할 때 `exam_score`(시험 점수)는 전반적으로 어떻게 변하는가? 두 변수 사이의 상관관계 유형을 단답형으로 쓰시오.

- **정답**: 양의 상관관계
- **해설**: 점들이 왼쪽 아래에서 오른쪽 위로 향하는 우상향 패턴을 그린다. 공부 시간이 길수록 시험 점수가 함께 높아지므로 두 변수는 양의 상관관계(상관계수 약 0.77)를 가진다.

### 문제 2

`hue='group'` 옵션으로 학습 방식을 색상으로 구분했을 때, 동일한 공부 시간 대비 시험 점수가 전반적으로 더 높게 분포하는 그룹은 무엇인가?

- **정답**: 온라인
- **해설**: 온라인 그룹의 점들이 오프라인 그룹보다 같은 x값(공부 시간)에서 더 위쪽(높은 점수)에 위치한다. 평균 점수도 온라인(약 68점)이 오프라인(약 59점)보다 높아, 온라인 그룹의 학습 효율이 더 높음을 알 수 있다.

### 문제 3

공부 시간이 많은 편(약 8.7시간)임에도 점수가 매우 낮아(약 20점) 전체 우상향 추세에서 크게 벗어나 있는 이상치(outlier)의 개수는 몇 개인가?

- **정답**: 1개
- **해설**: 산점도 오른쪽 아래(공부 시간은 길지만 점수는 낮은 영역)에 다른 점들과 동떨어진 점이 하나 보인다. 이 점은 전체적인 양의 관계 추세에서 벗어난 유일한 이상치로, 분석 시 별도로 검토해야 할 대상이다.
