## 문제 풀이

### 문제 1

2020년부터 2024년까지 매년 구성 비중이 한 번도 반등 없이 지속적으로 감소한 결제 수단은 무엇인가?

- **정답**: 현금
- **해설**: 현금의 구성 비중은 2020년 40% → 2021년 32% → 2022년 25% →
  2023년 18% → 2024년 10%로 매년 감소했다. 신용카드(40→43→45→44→42%)는
  중간에 증가·감소가 섞여 있고, 모바일페이는 증가, 상품권은 5→5→3→3→3%로
  중간에 변동이 없는 구간이 있으므로, 한 번의 반등도 없이 매년 감소한 것은
  현금뿐이다.

### 문제 2

모바일페이의 구성 비중이 신용카드의 구성 비중을 처음으로 추월한 연도는 언제인가?

- **정답**: 2024년
- **해설**: 연도별 (모바일페이 vs 신용카드) 비중은 2020년 15% vs 40%,
  2021년 20% vs 43%, 2022년 27% vs 45%, 2023년 35% vs 44%로 모두 신용카드가
  높았으나, 2024년에는 45% vs 42%로 모바일페이가 신용카드를 처음으로 추월한다.

### 문제 3

2020년 대비 2024년에 구성 비중의 증가폭(%)이 가장 큰 결제 수단은 무엇인가?

- **정답**: 모바일페이
- **해설**: 2020년→2024년 비중 변화는 현금 40%→10%(-30%p),
  신용카드 40%→42%(+2%p), 모바일페이 15%→45%(+30%p),
  상품권 5%→3%(-2%p)이다. 따라서 증가폭이 +30%p로 가장 큰 결제 수단은
  모바일페이다.
