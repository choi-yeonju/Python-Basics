### 문제 1

2015년과 2016년 모두 막대의 높이가 가장 낮게 나타나, 종사자 수가 가장 적은 종사자 규모 구간은?

- **정답**: 200 ~ 299명
- **해설**: 그래프에서 가장 낮은 막대 한 쌍은 200~299명 구간으로, 2015년 약 80.9, 2016년 약 81.0 수준이다. 이는 중소·중견기업 사이의 경계 구간으로, 다른 어떤 규모 구간보다도 종사자 분포가 적게 형성되어 있음을 의미한다.

### 문제 2

2015년에는 종사자 수 1위였으나, 2016년에는 1~4명 규모에게 1위를 빼앗긴 종사자 규모 구간은?

- **정답**: 1000명 이상
- **해설**: 2015년 기준 1000명 이상 구간은 약 557.6으로 가장 높았고, 1~4명 구간은 약 553.5로 근소하게 2위였다. 그러나 2016년에는 1~4명 구간이 576.7로 증가한 반면, 1000명 이상 구간은 549.7로 감소하면서 순위가 역전되었다. 즉 영세 사업장의 종사자 수가 늘고 초대규모 사업장 종사자 수는 줄어드는 변화가 관찰된다.

### 문제 3

2015년 대비 2016년에 종사자 수 증가폭이 가장 크게 나타난 종사자 규모 구간은?

- **정답**: 1 ~ 4명
- **해설**: 막대그래프에서 2015년과 2016년 두 막대의 높이 차이가 가장 크게 벌어지는 구간은 1~4명 구간으로, 약 553.5에서 576.7로 약 23 이상 증가하였다. 다른 구간들은 변화가 거의 없거나 소폭 감소(예: 1000명 이상 약 -7.9)하는 가운데, 영세 사업장(1~4명)에서만 두드러진 종사자 수 증가가 관찰되어 1년 사이 종사자 분포가 가장 작은 규모 쪽으로 이동하는 흐름을 보인다.
