## 연습문제 1

kings_life 데이터 셋은 조선시대 역대 왕의 수명을 조사한 데이터이다. 

수명에 대해 기술통계량을 구하고 상자그림을 그려라. 최종적으로 분석 결과를 토대로 하여 알 수 있는 사실을 하나 이상 제시하라.

### 1. 데이터 가져오기

```python
df = load_data("kings_life")
df.head()
```

### 2. 상자그림 그리기

```python
my_plot.boxplot(data=df, x='수명', xlabel="수명(세)", width=500, height=200)
```

## 문제 풀이

### 문제 1

상자그림(Box Plot)에서 조선시대 역대 왕들의 수명 중앙값은 약 몇 세인가?

- **정답**: 약 45세
- **해설**: 박스플롯에서 상자(box) 내부에 그려진 세로선이 중앙값(median, Q2)을 나타낸다. 시각화 결과에서 이 선은 약 45세 부근에 위치하고 있어, 조선시대 왕의 절반은 45세 이전에 사망했고 나머지 절반은 45세 이후에 사망했음을 알 수 있다.

### 문제 2

상자(box)의 좌우 경계로부터 계산되는 사분위범위(IQR)는 약 얼마인가?

- **정답**: 약 23세
- **해설**: 박스의 왼쪽 경계는 1사분위수(Q1 ≈ 33세), 오른쪽 경계는 3사분위수(Q3 ≈ 56세)에 해당하며, 두 값의 차이인 IQR은 약 23세이다. 이는 조선시대 왕의 중간 50%가 약 33세에서 56세 사이의 수명을 가졌음을 의미하며, 상당수의 왕이 60세를 넘기지 못하고 비교적 이른 나이에 사망했음을 보여준다.

### 문제 3

상자그림에서 수염(whisker) 바깥쪽에 표시된 점의 개수는 몇 개이며, 이를 통해 알 수 있는 사실은 무엇인가?

- **정답**: 0개 (이상치 없음)
- **해설**: 박스플롯에서 수염을 벗어난 점이 하나도 표시되지 않았으므로, 통계적 기준(Q1 - 1.5·IQR ~ Q3 + 1.5·IQR 바깥)으로 정의되는 이상치(outlier)는 존재하지 않는다. 또한 오른쪽 수염(Q3 ~ 최댓값 약 82세)이 왼쪽 수염(최솟값 약 16세 ~ Q1)보다 더 길게 나타나, 분포가 오른쪽으로 약간 치우친(우측 꼬리가 긴, positively skewed) 형태임을 함께 확인할 수 있다.


## 연습문제 2

stock 데이터셋은 상장된 금융사 중 외국기업(F)과 내국기업(K)의 주가 데이터의 일부이다.

외국기업과 내국기업별로 상자그림을 그려서 비교/분석하여 다음 물음에 답하시오.

### 1. 데이터 가져오기

```python
df = load_data("stock")
df.head()
```

### 2. 상자그림 그리기

```python
my_plot.boxplot(data=df, y='주가', ylabel="주가", hue='구분', width=300, height=300)
```

## 문제 풀이

### 문제 1

두 집단의 상자그림에서 중앙값(median)이 더 높은 집단은 외국기업(F)과 내국기업(K) 중 무엇인가?

- **정답**: K (내국기업)
- **해설**: 박스플롯에서 각 상자(box) 내부의 가로선이 중앙값을 나타낸다. F의 중앙값은 약 150 부근, K의 중앙값은 약 154 부근에 위치하고 있어 내국기업(K)의 주가 중앙값이 외국기업(F)보다 다소 높음을 알 수 있다. 즉 절반 이상의 내국기업이 외국기업의 중앙값보다 높은 주가를 형성하고 있다.

### 문제 2

상자(box)의 길이(=IQR, 사분위범위)를 통해 알 수 있는 주가의 변동성이 더 큰 집단은 무엇인가?

- **정답**: K (내국기업)
- **해설**: F의 IQR은 약 15(Q1 ≈ 143, Q3 ≈ 158), K의 IQR은 약 24(Q1 ≈ 142, Q3 ≈ 166)이다. K의 박스가 F의 박스보다 세로로 더 길게 표시되어 있어, 가운데 50%에 해당하는 데이터의 산포가 더 넓다. 따라서 내국기업(K)의 주가 변동 폭이 외국기업(F)보다 크다고 해석할 수 있다.

### 문제 3

상자그림에서 수염(whisker) 바깥쪽 점으로 표시된 이상치(outlier)의 총 개수는 몇 개인가?

- **정답**: 2개 (F 1개, K 1개)
- **해설**: 외국기업(F)에는 박스 위쪽에 약 235 부근에 점으로 표시된 이상치가 1개 존재하며, 내국기업(K)에는 박스 아래쪽에 약 102 부근에 점으로 표시된 이상치가 1개 존재한다. 즉 F는 비정상적으로 주가가 높은 기업, K는 비정상적으로 주가가 낮은 기업이 각각 한 개씩 관찰되어 총 2개의 이상치가 발견된다.

## 문제 풀이

### 문제 1

두 비료 집단의 상자그림에서 수확량의 중앙값(median)이 더 높은 비료는 A와 B 중 무엇인가?

- **정답**: A

- **정답**: B

### 문제 3

상자그림에서 수염(whisker) 바깥쪽 점으로 표시된 이상치(outlier)가 발견된 비료와 그 개수는?

- **정답**: A, 2개
- **해설**: 비료 A에는 박스 위쪽 약 39.3 부근과 박스 아래쪽 약 17.8 부근에 각각 점으로 표시된 이상치가 1개씩, 총 2개가 존재한다. 반면 비료 B는 수염 바깥쪽에 점이 없어 이상치가 발견되지 않는다. 즉 비료 A는 대체로 안정적인 수확량을 보이지만 일부 시료에서 비정상적으로 높거나 낮은 수확량이 관찰되었음을 알 수 있다.

## 문제 풀이

### 문제 1

상자그림에서 이상치까지 포함한 전체 관측값을 기준으로, 가장 높은 점수를 받은 학생이 속한 학과는?

- **정답**: C
- **해설**: 학과 A의 위쪽 수염 끝(최댓값)은 약 78점이지만, 학과 C에는 박스 위쪽에 약 90점과 약 82점에 점으로 표시된 이상치가 존재한다. 즉 학과 C의 박스/수염 자체는 좁게 형성되어 있어 일반적인 점수대는 50점대 초반에 집중되어 있지만, 예외적으로 매우 높은 점수를 받은 학생이 있어 전체 데이터에서 최고점(약 90점)은 학과 C에서 관측된다. 박스 본체만 보고 비교하면 놓치기 쉬운 인사이트이다.

### 문제 2

두 학과의 박스(Q1~Q3 구간)가 서로 공통으로 겹치는 점수 구간은 약 몇 점부터 몇 점까지인가?

- **정답**: 약 51점 ~ 55점
- **해설**: 학과 C의 박스는 약 51~54점, 학과 A의 박스는 약 30~55점 구간에 형성되어 있다. 두 박스가 공통으로 겹치는 구간은 약 51점에서 55점 사이로, 학과 C의 박스가 거의 통째로 학과 A의 박스 상단 부근에 포함되는 형태이다. 이는 학과 C의 가운데 50%에 해당하는 학생들이 학과 A의 상위권 점수대(상위 25% 부근)와 비슷한 분포를 보임을 의미한다.

### 문제 3

학과 C의 박스(Q1~Q3) 기준 위쪽 방향과 아래쪽 방향에서 각각 관측된 이상치의 개수는?

- **정답**: 위쪽 2개, 아래쪽 2개
- **해설**: 학과 C에는 박스 위쪽으로 약 82점과 약 90점 부근에 이상치가 각각 1개씩(총 2개), 박스 아래쪽으로 약 41점과 약 43점 부근에 이상치가 각각 1개씩(총 2개) 나타난다. 단순히 "이상치가 존재한다"는 사실을 넘어서, C의 이상치가 위·아래 양방향으로 균등하게 발생하고 있다는 점이 특징이며, 이는 대다수가 50점대 초중반에 밀집된 가운데 양극단의 예외적 학생들이 모두 존재함을 보여준다.
