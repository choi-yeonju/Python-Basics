## #02. 요구사항 해결

1. 분석에 필요한 `Flavor`(맛)와 `Quantity`(판매 수량) 두 컬럼만 선택한다.
2. `Flavor`를 기준으로 그룹화한 뒤 `Quantity`의 합계를 계산하여 맛별 총 판매량 데이터를 만든다.
3. 맛별 총 판매량을 내림차순으로 정렬한다.
4. 정렬된 데이터를 사용해 파이 차트를 그린다. (값: `Quantity`, 라벨: 맛 이름)
5. `palette` 옵션을 지정하여 시각화를 완성한다.

### 1. 데이터 전처리

```python
df1 = origin[['Flavor', 'Quantity']].groupby('Flavor').sum()
df1.sort_values('Quantity', ascending=False, inplace=True)
df1.head()
```

### 2. 파이 그래프

```python
my_plot.pieplot(x=df1['Quantity'], labels=df1.index, palette="pastel")
```

## 문제 풀이

### 문제 1

파이 차트에서 가장 큰 비중을 차지하여 판매량 1위로 나타난 아이스크림 맛은?

- **정답**: 바닐라
- **해설**: 파이 차트에서 가장 넓은 면적(34.6%)을 차지하는 조각은 바닐라이며, 총 83개로 4가지 맛 중 가장 많이 팔렸다. 이는 매장의 대표 인기 메뉴가 바닐라임을 시각적으로 한눈에 보여준다.

### 문제 2

파이 차트에서 가장 작은 비중으로 표시되어 판매량이 가장 적은 아이스크림 맛은?

- **정답**: 민트초코
- **해설**: 민트초코 조각은 전체의 13.3%로 가장 작은 면적을 차지하며, 판매량은 32개로 다른 맛 대비 절반에 못 미치는 수준이다. 즉 호불호가 갈리는 맛 특성이 판매 비중에도 그대로 반영되어, 매장 내에서 가장 인기가 낮은 메뉴로 확인된다.

### 문제 3

파이 차트에서 점유율 합계가 전체의 절반(50%)을 넘는 상위 두 맛은?

- **정답**: 바닐라, 딸기
- **해설**: 1위인 바닐라(34.6%)와 2위인 딸기(27.1%)의 점유율 합은 약 61.7%로 전체의 절반을 훌쩍 넘는다. 즉 이 두 맛이 매장 매출의 핵심을 담당하고 있으며, 재고·생산 우선순위 측면에서 이 두 메뉴를 중심으로 운영 전략을 세울 필요가 있음을 시사한다.
