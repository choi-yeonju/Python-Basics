## #03. 문제 풀이

### 문제 1

통계학 전공의 중간고사 점수 분포에 나타난 봉우리(peak)는 몇 개인가?

- **정답**: 2개
- **해설**: 약 45점 부근과 90점 부근에 각각 봉우리가 있는 이봉형(bimodal) 분포이다. 학생들이 상위권과 하위권의 두 집단으로 양극화되어 있음을 의미한다.

### 문제 2

통계학 전공의 중간고사 분포에서 학생 수가 0명인 점수 구간은 약 몇 점대인가?

- **정답**: 약 54~64점 (60점대 초반)
- **해설**: 두 봉우리 사이의 골짜기에 해당하는 구간으로, 중간 수준의 점수를 받은 학생이 한 명도 없음을 보여준다. 양극화의 직접적인 증거이다.

### 문제 3

통계학 전공 학생들의 기말고사에서 가장 많은 학생이 분포한 점수 구간은?

- **정답**: 약 74~88점
- **해설**: 기말고사 히스토그램에서 빈도 3명짜리 막대 두 개가 약 7481점, 8188점 구간에 연속으로 나타난다. 중간고사에서 양극화되었던 분포가 기말에는 고득점대로 수렴함을 보여준다.

### 문제 4

전산학 전공의 기말고사 점수 분포의 왜도 방향은? (양/음 중 택1)

- **정답**: 음의 왜도 (좌측 꼬리)
- 해설: 8090점대에 학생이 가장 많이 몰려 있고, 4050점대 쪽으로 길게 꼬리가 형성된다. `최빈값 < 중앙값 < 평균`이 아니라 그 반대(`평균 < 중앙값 < 최빈값`) 관계가 성립한다.

### 문제 5

두 전공의 중간고사와 기말고사를 비교했을 때, 학생들의 점수는 전반적으로 (높아졌다 / 낮아졌다) 중 어느 쪽인가?

- **정답**: 높아졌다 (상승)
- **해설**: 통계학 전공은 최저 점수가 34점 → 53점으로, 전산학 전공은 30점 → 40점으로 이동했고, 두 전공 모두 기말에는 70점 이상 구간에 학생이 집중되었다. 한 학기 동안 학업 성취도가 향상되었음을 시사한다.
