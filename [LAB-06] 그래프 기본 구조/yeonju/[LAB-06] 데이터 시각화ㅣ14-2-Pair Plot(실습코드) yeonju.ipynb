### 14-2 Pair Plot 

### 1. 라이브러리 참조 

```python
from jussam import load_data
from helpers import my_plot
import seaborn as sb
```

### 2. 데이터 가져오기 

```python
origin = load_data('penguins')
origin
```

```python
# 1) 그래프 초기화 
width = 1600 # 그래프 가로 px 크기 
height = 1200 # 그래프 세로 px 크기 
my_dpi = 200      # 그래프 해상도 (DPI)

# 그래프 크기를 inch 단위로 변환 
figsize = (width / 100, height / 100)
```

```python
# 2) Pait Plot 그리기 
g = sb.pairplot(origin)
g.fig.set_dpi(my_dpi)
g.fig.set_figwidth(figsize[0])
g.fig.set_figheight(figsize[1])

# 제목 추가 
g.fig.suptitle("펭귄 데이터의 Pair Plot", fontsize=16, fontweight="bold")
```

### 1. 기본형 (2/2)

```python
# 3) 개별 그래프 설정 
for ax in g.axes.flatten():
    # 격자를 그래프 뒤로 이동
    ax.set_axisbelow(True)
    # 격자 추가 
    ax.grid(True, alpha=0.5)

# 4) 출력 
my_plot.show()
```

### 2. 범주별 구분 및 마커 모양 설정 (1/2)

```python
# 1) 그래프 초기화 
width = 1600 # 그래프 가로 px 크기 
height = 1200 # 그래프 세로 px 크기 
my_dpi = 200      # 그래프 해상도 (DPI)

# 그래프 크기를 inch 단위로 변환 
figsize = (width / 100, height / 100)
```

```python
# 2) Pait Plot 그리기 
g = sb.pairplot(origin, 
                hue='species', 
                markers=['o', 's', '^'], 
                palette='cool', 
                plot_kws={'s':30, 'alpha': 0.7}
                )
```

```python
g.fig.set_dpi(my_dpi)
g.fig.set_figwidth(figsize[0])
g.fig.set_figheight(figsize[1])

# 제목 추가 
g.fig.suptitle("펭귄 데이터의 Pair Plot", fontsize=16, fontweight="bold")
```

```python
# 3) 개별 그래프 설정 
for ax in g.axes.flatten():
    # 격자를 그래프 뒤로 이동
    ax.set_axisbelow(True)
    # 격자 추가 
    ax.grid(True, alpha=0.5)

# 4) 출력 
my_plot.show()
```

### 3. 회귀선 추가하기 

```python
# 1) 그래프 초기화 
width = 1600 # 그래프 가로 px 크기 
height = 1200 # 그래프 세로 px 크기 
my_dpi = 200      # 그래프 해상도 (DPI)

# 그래프 크기를 inch 단위로 변환 
figsize = (width / 100, height / 100)
```

```python
# 2) Pait Plot 그리기 
g = sb.pairplot(origin, 
                hue='species', 
                markers=['o', 's', '^'], 
                palette='cool', 
                diag_kind ='kde', 
                kind='reg',
                # 회귀선을 설정한 경우 모양 설정 방법이 세분화 됨 
                plot_kws={
                    # 마커 설정 
                    'scatter_kws': {'s':15, 'alpha':0.5}, 
                    # 회귀선 스타일 
                    'line_kws' : {'linewidth':2, 'linestyle':'--'}  
                    },
                )
```

```python
# 4) 출력 
my_plot.show()
```

### 4. 선택적 변수 적용 

```python
# 1) 그래프 초기화 
width = 1600 # 그래프 가로 px 크기 
height = 1200 # 그래프 세로 px 크기 
my_dpi = 200      # 그래프 해상도 (DPI)

# 그래프 크기를 inch 단위로 변환 
figsize = (width / 100, height / 100)
```

```python
# 2) Pait Plot 그리기 
g = sb.pairplot(origin, 
                hue='species', 
                diag_kind ='kde', 
                kind='reg',
                x_vars=["bill_length_mm", "bill_depth_mm"], 
                y_vars=["bill_length_mm", "bill_depth_mm"]
                )

g.fig.set_dpi(my_dpi)
g.fig.set_figwidth(figsize[0])
g.fig.set_figheight(figsize[1])
```

```python
# 4) 출력 
my_plot.show();
```
