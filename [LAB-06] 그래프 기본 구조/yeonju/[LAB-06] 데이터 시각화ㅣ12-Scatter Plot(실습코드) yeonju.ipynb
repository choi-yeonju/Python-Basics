## #02. Scatter Plot 시각화 

### 1. 기본 사용 방법 - 팽귄의 날개 길이와 몸무게의 관계 

```python
# 1) 그래프 초기화 
my_plot.init(title="팽귄의 날개 길이와 몸무게의 관계", xlabel="날개 길이", ylabel="몸무게")

# 2) Scatter Plot 그리기 
# 패턴 1 - 자동인식 패턴 => 모든 컬럼에 대해 작용하므로 권장하지 않음 
# sb.scatterplot (data=origin)
# 패턴 2 - 개별 데이터 직접 지정 패턴 => 사용 가능 
# sb.scatterplot (x=origin['flipper_length_mm'], y=origin['body_mass_g'])
# 패턴 3 - DataFrame + 컬럼명 패턴 => 가장 권장하는 패턴 
sb.scatterplot( data = origin, x='flipper_length_mm', y='body_mass_g')

# 3) 그래프 출력 
my_plot.show()
```

### 2. Scatter Plot의 옵션 - 팽귄의 날개 길이와 몸무게의 관계 

```python
# 1) 그래프 초기화 
my_plot.init(title="펭귄의 날개 길이와 몸무게의 관계", xlabel="날개 길이", ylabel="몸무게")

# 2) Scatter Plot 그리기 
sb.scatterplot(data=origin, x='flipper_length_mm', y='body_mass_g', 
               markers="^",         # 마커 모양 
               color="red",         # 마커 색상 
               s = 150,             # 마커 크기 (기본값=100)
               edgecolor="black",   # 마커 테두리 색상 
               linewidth=2,         # 마커 테두리 두께 
               alpha=0.7            # 마커 투명도 (1=불투명, 0=투명)
               )
# 3) 그래프 출력 
my_plot.show()
```

### 3. 군집별 분류 - 펭귄 종에 따른 부리 길이와 부리 두께의 관계 

```python
# 1) 그래프 초기화 
my_plot.init(title="펭귄의 부리 길이와 부리 두께의 관계", 
             xlabel="부리 길이", ylabel="부리 두께")

# 2) Scatter Plot 그리기 
sb.scatterplot(data=origin, x='bill_length_mm', y='bill_depth_mm', 
               hue="species",       # 마커 색상을 종(species)별로 다르게 설정 
               palette = "tab10",   # 색상 팔레트 설정 
               markers="o",         # 마커 모양 
               s = 100,             # 마커 크기 (기본값=100)
               edgecolor="black",   # 마커 테두리 색상 
               linewidth=2,         # 마커 테두리 두께 
               alpha=0.7            # 마커 투명도 (1=불투명, 0=투명)
               )
# 3) 그래프 출력 
my_plot.show()
```
