## 문제 2

`salary` 데이터셋은 어느 회사 직원들의 **근속 년수(YearsExperience)** 와 **연봉(Salary)** 을 조사한 자료입니다.

이 회사 인사팀에 입사한 당신은 근속 년수에 따른 연봉 테이블을 만들기 위해, **추론통계 기반 탐색적 데이터분석 → 단순선형회귀 모델 적합 → 회귀분석의 가정 점검 → 결과 보고**로 이어지는 전 과정을 수행해야 합니다.

### 데이터 가져오기

```python
df = load_data("salary")
df.head()
```

### Q1. `salary` 데이터의 두 변수(`YearsExperience`, `Salary`)에 대해 **결측치 · 이상치 존재 여부**와 **로그 변환 필요 여부**를 판단하시오.

```python
# 결측치 점검
display(my_qtcheck.check_missing_values(df))

# 기술통계 + 이상치 / 왜도 / 로그변환 판단
my_qtcheck.numerical_summary(df).T
```

> **정답 (Q1)** — 결측치 **0개** · IQR 이상치 **0개** · **로그 변환 불필요**(두 변수 모두 |왜도| < 1, `log_need = none`)

### Q2. `근속 년수`와 `연봉`의 **상관분석**을 수행했을 때 자동으로 선택되는 상관계수의 **종류·값**, 그리고 **선택 이유**와 **유의성**을 답하시오.

```python
my_stats.correlation(df, x="YearsExperience", y="Salary").T
```

> **정답 (Q2)** — **Spearman ρ ≈ 0.957** (유의, 강한 양의 상관). `Salary`가 정규성을 만족하지 않아 Pearson 대신 선택

### Q3. `Salary`를 종속변수, `YearsExperience`를 독립변수로 하는 **단순선형회귀식**을 절편과 기울기를 포함하여 도출하시오.

```python
fit = my_ols.fit_model(df, "Salary", summary=True)
```

> **정답 (Q3)** — **Salary = 24,848.20 + 9,449.96 × YearsExperience**

### Q4. 이 모형의 **결정계수(R²)** 값과 그 의미, 그리고 **모형의 통계적 유의성**을 답하시오.

```python
Markdown(my_ols.report_fitness(fit))
```

> **정답 (Q4)** — **R² ≈ 0.957** (연봉 분산의 약 95.7% 설명), **F(1, 28) = 622.51, p < .001 → 모형 유의**

### Q5. 근속 년수의 **회귀계수(B)** 값과 그 **실질적 의미**를 서술하시오.

```python
my_ols.report_variables(fit, df)
```

> **정답 (Q5)** — **B ≈ 9,449.96** → 근속 년수 1년 증가 시 연봉 평균 약 **9,450 증가**

### Q6. 적합된 모형으로 **근속 3·5·7·10년** 직원의 예측 연봉을 각각 구하고, 특히 **근속 7년**의 예측 연봉을 답하시오.

```python
new_data = pd.DataFrame({"YearsExperience": [3, 5, 7, 10]})
my_ols.predict(fit, new_data)
```

> **정답 (Q6)** — 근속 **7년 ≈ 90,998** (3년 53,198 · 5년 72,098 · 10년 119,348)

### Q7. 이 모형의 **잔차 정규성**을 검정할 때 사용되는 **검정 방법**과 그 **결과(충족 여부)** 를 답하고, 왜 그 검정 방법이 사용되는지 근거를 쓰시오.

```python
my_ols.test_normal(fit)
```

> **정답 (Q7)** — **Kolmogorov-Smirnov** 검정(∵ n = 30 ≥ 30), **p ≈ 0.974 → 정규성 만족**

### **Q8.** 회귀분석의 **네 가지 가정(선형성 · 정규성 · 등분산성 · 독립성)**의 **충족 여부**를 종합하여 답하시오.

```python
# 선형성 (Ramsey RESET)
my_ols.test_linear(fit)
```

```python
# 정규성
my_ols.test_normal(fit)
```

```python
# 등분산성 (Breusch-Pagan)
my_ols.test_equalvar(fit)
```

```python
# 독립성 (Durbin-Watson)
my_ols.test_independent(fit)
```

> **정답 (Q8)** — **네 가정 모두 충족** (선형성 p ≥ 0.05 · 정규성 p ≥ 0.05 · 등분산 p ≥ 0.05 · 독립성 DW ≈ 1.65)
